# Insights and observation
The dataset has missing values

```python
#check if all the values in number of the reviews
df['Reviews'].unique()
```

```python
#as the reviews is in object formate so i am checking that whether there is any numeric value or not
df['Reviews'].str.isnumeric().sum()
```

```python
#not(~) 
df[~df['Reviews'].str.isnumeric()]
```

```python
df_copy= df.copy()
df_copy= df_copy.drop(df_copy.index[10472])
```

```python
df_copy[~df_copy['Reviews'].str.isnumeric()]
```

```python
#convet the reviews datatype to int
df_copy['Reviews']= df_copy['Reviews'].astype(int)
```

```python
df_copy.info()
```

```python
df_copy['Size'].unique()
```

```python
# 19000k == 19M
```

```python
df_copy['Size'].isnull().sum()
```

```python
df_copy['Size']= df_copy['Size'].str.replace('M','000')
df_copy['Size']= df_copy['Size'].str.replace('k','')
df_copy['Size']= df_copy['Size'].replace('Varies with device',np.nan)

df_copy['Size']= df_copy['Size'].astype(float)
```

```python
df_copy.info()
```

# Handling Price and Installs features

```python
df_copy['Installs'].unique()
```

```python
df_copy['Price'].unique()
```

```python
chars_to_remove=['+',',', '$']
cols_to_clean=['Installs', 'Price']
for item in chars_to_remove:
    for cols in cols_to_clean:
        df_copy[cols] = df_copy[cols].str.replace(item,'')
        
```

```python
df_copy['Price'].unique()
```

```python
df_copy['Installs'].unique()
```

```python
df_copy['Installs'] = df_copy['Installs'].astype(int)
df_copy['Price']= df_copy['Price'].astype(float)
```

```python
df_copy.info()
```

# Handling last update features


```python
 df_copy['Last Updated'].unique()
```

```python
df_copy['Last Updated']=pd.to_datetime(df_copy['Last Updated'])
```

```python
df_copy['Day']=df_copy['Last Updated'].dt.day
df_copy['Month']=df_copy['Last Updated'].dt.month
df_copy['Year']=df_copy['Last Updated'].dt.year
```

```python
df_copy.info()

```

```python
df_copy.drop('Last Updated', axis=1, inplace=True)
```

```python
df_copy.info()
```

```python
df_copy.to_csv('data/google_cleaned.csv')
```

# Explore data more

```python
numeric_features     =[feature for feature in df_copy.columns if df_copy[feature].dtype !='O']
categorical_features =[feature for feature in df_copy.columns if df_copy[feature].dtype =='O']

#print columns
print("We have {} numerical features: {}".format(len(numeric_features),numeric_features))
print("\nWe have {} categorical features : {}".format(len(categorical_features), categorical_features))
```

## 3.2 Feature Information

1. **App** :- Name of the App  
2. **Category** :- Category under which the App falls.  
3. **Rating** :- Application's rating on Play Store  
4. **Reviews** :- Number of reviews of the App.  
5. **Size** :- Size of the App.  
6. **Install** :- Number of Installs of the App  
7. **Type** :- If the App is free/paid  
8. **Price** :- Price of the app (0 if it is Free)  
9. **Content Rating** :- Appropriate Target Audience of the App.  
10. **Genres** :- Genre under which the App falls.  
11. **Last Updated** :- Date when the App was last updated  
12. **Current Ver** :- Current Version of the Application  
13. **Android Ver** :- Minimum Android Version required to run the App

```python
#Proportion of count data on categoriacal columns
for col in categorical_features:
    print(df[col].value_counts(normalize=True)*100)
    print("---------------------------------------")
```

```python
# Proportion of count data on numerical columns

plt.figure(figsize=(15, 15))
plt.suptitle(
    "Univariate Analysis of Numerical Features",
    fontsize=20,
    fontweight='bold',
    alpha=0.8,
    y=1
)

for i, feature in enumerate(numeric_features):
    plt.subplot(5, 3, i + 1)
    sns.kdeplot(
        x=df_copy[feature],
        fill=True
    )
    plt.xlabel(feature)
    plt.tight_layout()

plt.show()
```

# Obeservation
- Rating and year is left skewed and while reviews,size,installs and price are rigth skewed

```python
# Proportion of count data on categorical columns

plt.figure(figsize=(20, 15))
plt.suptitle(
    "Univariate Analysis of Categorical Features",
    fontsize=20,
    fontweight='bold',
    alpha=0.8,
    y=1
)
category= ['Type','Content Rating']

for i, feature in enumerate(category):
    plt.subplot(2, 2, i + 1)
    sns.countplot(x=df[category[i]], palette="Set2")
    plt.xlabel(category[i])
    plt.xticks(rotation=45)
    plt.tight_layout()

plt.show()
```

```python
## Top 10 App categories
category=pd.DataFrame(df['Category'].value_counts()) #Dataframe of aps on the basis of category
category.rename(columns={'Category': 'Count'},inplace=True)
```

```python
category
```

```python
plt.figure(figsize=(15, 6))

sns.barplot(
    x=category.index[:10],
    y=category['count'][:10]
)

plt.title("Top 10 App Categories")
plt.xticks(rotation=90)
plt.show()
```

## Obeservations
1. Family category has the most number of apps with 18% of apps belonging to it, followed by games category which has 11% if the apss
2. Least number of apps belongs to the beauty category with less than 1% of the total apps belonging to it
