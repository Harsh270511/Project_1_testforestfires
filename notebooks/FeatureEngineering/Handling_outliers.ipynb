# 5 number summary and box plot

```python
# Minimum, Maximum, Q1, Q2, Median, IQR
```

```python
import numpy as np
```

```python
lst_marks=[45,32,56,75,89,54,32,89,90,87,67,54,45,98,99,67,79]
minimum, Q1, Median, Q3, maximum= np.quantile(lst_marks, [0, 0.25, 0.50, 0.75, 1])
```

```python
minimum,Q1, Median, Q3, maximum 
```

```python
IQR= Q3 - Q1
print(IQR)
```

```python
lower_fence= Q1-1.5*IQR
print(lower_fence)
```

```python
upper_fence= Q3 + 1.5*IQR
print(upper_fence)
```
